```
#| export
def load_data(src, **kwargs):
    """Load provider-specific input data for the handler template."""

    return {
        "src": Path(src),
        "provider_options": dict(kwargs),
    }
```

## Transformation pipeline

```
#| export
def build_pipeline(callbacks: Optional[Iterable[Any]] = None):
    """Return the ordered callback pipeline for a handler implementation."""

    return list(callbacks) if callbacks is not None else [TemplateCB()]
```

## Encode

The exported `encode` entrypoint stays named `encode` for CLI and handler consistency. Replace the placeholder return payload with the concrete NetCDF or intermediate export path that the real handler needs.

```
#| export
def encode(fname_in, fname_out, **kwargs):
    """Template encode entrypoint for notebook-authored handlers."""

    loaded = load_data(fname_in, **kwargs)
    return {
        "fname_in": Path(fname_in),
        "fname_out": Path(fname_out),
        "attrs": get_attrs(),
        "pipeline": build_pipeline(),
        "loaded": loaded,
    }
```
